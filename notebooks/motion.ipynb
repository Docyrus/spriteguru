```python
import json, numpy as np
from pathlib import Path
from PIL import Image
from IPython.display import display
ROOT = Path.cwd().parent
def show(a, scale=1):
    im = Image.fromarray(a if a.dtype == np.uint8 else (np.clip(a, 0, 1) * 255).astype(np.uint8))
    if scale != 1: im = im.resize((im.width * scale, im.height * scale), Image.NEAREST)
    display(im)
def fixture(case):
    d = ROOT / 'tests' / 'golden' / case
    from spriteguru.pipeline.matte import ingest
    from spriteguru.spec import SpriteSpec
    t = json.loads((d / 'truth.json').read_text())
    return ingest(d / 'sheet.png'), SpriteSpec.model_validate(t['spec']), t
```

```python
from spriteguru.pipeline import motion, vector
from spriteguru.figure import character_skin
svg = vector.synthetic_character_svg(character_skin(2), 256, 256)
rig = motion.load_rig(svg)
print(motion.validate_rig(rig), rig.pivots)
```

```python
from types import SimpleNamespace
spec = motion.synthetic_author(SimpleNamespace(params={'action': 'walk', 'frames': 8, 'width': 256, 'height': 256}))
print(motion.validate_spec(spec, rig, loop=True))
import asyncio
frames = asyncio.run(motion.render_spec(spec, rig, 8, True))
from spriteguru.report import contact_sheet
display(contact_sheet(frames))
```
