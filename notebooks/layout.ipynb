Lines, labels, the three detectors and their scores.

```python
import json, numpy as np
from pathlib import Path
from PIL import Image
from IPython.display import display
ROOT = Path.cwd().parent
def show(a, scale=1):
    im = Image.fromarray(a if a.dtype == np.uint8 else (np.clip(a, 0, 1) * 255).astype(np.uint8))
    if scale != 1: im = im.resize((im.width * scale, im.height * scale), Image.NEAREST)
    display(im)
def fixture(case):
    d = ROOT / 'tests' / 'golden' / case
    from spriteguru.pipeline.matte import ingest
    from spriteguru.spec import SpriteSpec
    t = json.loads((d / 'truth.json').read_text())
    return ingest(d / 'sheet.png'), SpriteSpec.model_validate(t['spec']), t
```

```python
rgba, spec, truth = fixture('grid-lines-labels')
from spriteguru.pipeline.matte import key_matte
from spriteguru.pipeline.layout import detect, Prior
m = key_matte(rgba)
lay = detect(m.mask, Prior(n=spec.frames))
print(lay.hypothesis, lay.confidence, lay.lines, lay.labels_removed)
print(json.dumps(lay.scores, indent=1))
```
